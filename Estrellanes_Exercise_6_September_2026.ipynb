```python
def Calc_Area_Circle():

  # Global Variables
  rad = float(input("Enter Radius: "))
  unit_of_measure = input("Enter unit of measurement: ")

  # Formula for the area of the circle
  circ_area = 3.14 * (rad ** 2)
  # Returns area of the circle and the unit of measure
  print("The Area of your circle is: ", str(circ_area) + " " + unit_of_measure)

Calc_Area_Circle()
```

```python
def login():
    # Correct Credentials
    correct_username = "Alvin"
    correct_password = "programming123"

    # User Input
    user = input("Enter username: ")
    password = input("Enter password: ")

    # Practice if, elif, else to check input
    if user != correct_username and password != correct_password:
        print("username and password are wrong")
    elif user != correct_username:
        print("username is wrong")
    elif password != correct_password:
        print("password is wrong")
    else:
        print("Welcome " + user)

login()
```

```python

def div():

  # User Inputs
  input1 = float(input("Enter first number (Dividend): "))
  input2 = float(input("Enter second number (Divisor): "))

  # Check if input2 is zero
  if input2 == 0:
    print("Undefine, please do not input zero (0) as Divisor")
  else:
    print("The Quotient is: ", input1 / input2)

div()



```
